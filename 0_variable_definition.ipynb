```python
import pandas as pd

# Read China dataset
df = pd.read_csv(
    "../data/raw/china.csv",
    low_memory=False
)

# 0. Outcome variables
target_vars = [
    "i10_health",   # perceived feasibility of self-isolation
    "i11_health"    # willingness to self-isolate
]

# 1. Protective behaviour predictors
behaviour_vars = (
    [f"i12_health_{i}" for i in range(1, 9)] +
    [f"i12_health_{i}" for i in range(11, 21)]
)

# 2. Demographic predictors
demographic_vars = [
    "gender",
    "age"
]

# 3. Household predictors
household_vars = [
    "household_size",
    "household_children"
]

# 4. Employment predictors
employment_vars = [
    f"employment_status_{i}" for i in range(1, 8)
]

# 5. Contact predictors
contact_vars = [
    "i1_health",
    "i2_health",
    "i7a_health"
]

# 6. Testing COVID
testing_vars = [
    "i3_health",
    "i4_health"
]

# 7. COVID-like symptom items
symptom_vars = [
    "i5_health_1",
    "i5_health_2",
    "i5_health_3",
    "i5_health_4",
    "i5_health_5",
    "i5_health_99"
]

# 8. Comorbidity items
comorbidity_vars = (
    [f"d1_health_{i}" for i in range(1, 14)] +
    ["d1_health_98", "d1_health_99"]
)

# 9. Time
time_vars = [
    "qweek"
]

# 10. Combine all selected variables
selected_vars = (
    target_vars
    + behaviour_vars
    + demographic_vars
    + household_vars
    + employment_vars
    + contact_vars
    + testing_vars
    + symptom_vars
    + comorbidity_vars
    + time_vars
)

# Remove duplicates
selected_vars = list(dict.fromkeys(selected_vars))

# Variables meeting the missingness criterion but excluded
excluded_vars = [
    "RecordNo",
    "endtime",
    "city",
    "weight",
    "i9_health",
    "i13_health"
]

# Check whether any expected variables are absent
missing_vars = [col for col in selected_vars if col not in df.columns]

if missing_vars:
    print("Variables not found:")
    print(missing_vars)

# Keep only variables that exist
selected_vars = [col for col in selected_vars if col in df.columns]

# Create project dataset
df_project = df[selected_vars].copy()

print("df_project shape:", df_project.shape)
print("\nSelected columns:")
print(df_project.columns.tolist())
```

```python
# Save selected raw variables for later processing
df_project.to_csv(
    "../data/data_processed/df_project_raw.csv",
    index=False
)
```
