```python
import pandas as pd

# Read the original China dataset
df_china = pd.read_csv(
    "../data/raw/china.csv",
    na_values=[" ", "__NA__"],
    keep_default_na=True,
    low_memory=False
)

# Calculate missing values for each variable
missing_summary = pd.DataFrame({
    "Variable": df_china.columns,
    "Missing_count": df_china.isna().sum().values,
    "Missing_percentage": (
        df_china.isna().mean().values * 100
    )
})

# Sort variables by missing percentage from lowest to highest
missing_summary = (
    missing_summary
    .sort_values(
        by=["Missing_percentage", "Variable"],
        ascending=[True, True]
    )
    .reset_index(drop=True)
)

# Round percentages for display
missing_summary["Missing_percentage"] = (
    missing_summary["Missing_percentage"].round(2)
)

print("Dataset shape:", df_china.shape)
print("\nMissing values by variable:")
print(missing_summary.to_string(index=False))
```

```python
# Identify variables with less than 10% missing values
variables_below_10_missing = missing_summary.loc[
    missing_summary["Missing_percentage"] < 10,
    "Variable"
].tolist()

# Print variables with less than 10% missing values
print("\nNumber of variables with less than 10% missing values:")
print(len(variables_below_10_missing))

print("\nVariables with less than 10% missing values:")
for variable in variables_below_10_missing:
    print(variable)
```
