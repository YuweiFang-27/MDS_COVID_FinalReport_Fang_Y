```python
import pandas as pd
from sklearn.model_selection import train_test_split

# Read model-ready dataset
df_model_ready = pd.read_csv("../data/data_processed/df_model_ready.csv")

# Create joint stratification variable
stratify_group = (
    df_model_ready["feasibility_binary"].astype(str)
    + "_"
    + df_model_ready["willingness_binary"].astype(str)
)

# Split into training and test sets
df_train, df_test = train_test_split(
    df_model_ready,
    test_size=0.2,
    random_state=42,
    stratify=stratify_group
)

print("Training set shape:", df_train.shape)
print("Test set shape:", df_test.shape)

# Save split datasets
df_train.to_csv("../data/data_processed/df_train.csv", index=False)
df_test.to_csv("../data/data_processed/df_test.csv", index=False)

# Check outcome distributions
for target_col in ["feasibility_binary", "willingness_binary"]:
    print(f"\n{target_col}")
    print("\nFull dataset:")
    print(df_model_ready[target_col].value_counts(normalize=True).sort_index().mul(100).round(2))

    print("\nTraining set:")
    print(df_train[target_col].value_counts(normalize=True).sort_index().mul(100).round(2))

    print("\nTest set:")
    print(df_test[target_col].value_counts(normalize=True).sort_index().mul(100).round(2))

# Check joint outcome distribution
print("\nFull dataset joint distribution:")
print(pd.crosstab(
    df_model_ready["feasibility_binary"],
    df_model_ready["willingness_binary"],
    normalize="all"
).mul(100).round(2))

print("\nTraining set joint distribution:")
print(pd.crosstab(
    df_train["feasibility_binary"],
    df_train["willingness_binary"],
    normalize="all"
).mul(100).round(2))

print("\nTest set joint distribution:")
print(pd.crosstab(
    df_test["feasibility_binary"],
    df_test["willingness_binary"],
    normalize="all"
).mul(100).round(2))
```
