```python
import pandas as pd
import os

# Read split datasets
df_train = pd.read_csv("../data/data_processed/df_train.csv")
df_test = pd.read_csv("../data/data_processed/df_test.csv")

# Define outcome variables
feasibility_target = "feasibility_binary"
willingness_target = "willingness_binary"

# Define predictor variables
feature_cols = df_train.columns.drop([
    feasibility_target,
    willingness_target
])

# Create X for both models
X_train = df_train[feature_cols]
X_test = df_test[feature_cols]

# Create y for feasibility model
y_train_feasibility = df_train[feasibility_target]
y_test_feasibility = df_test[feasibility_target]

# Create y for willingness model
y_train_willingness = df_train[willingness_target]
y_test_willingness = df_test[willingness_target]

# Check shapes
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)

print("\nFeasibility model:")
print("y_train_feasibility:", y_train_feasibility.shape)
print("y_test_feasibility:", y_test_feasibility.shape)

print("\nWillingness model:")
print("y_train_willingness:", y_train_willingness.shape)
print("y_test_willingness:", y_test_willingness.shape)

# Check outcome distributions in training data
print("\nFeasibility training distribution:")
print(y_train_feasibility.value_counts(normalize=True).sort_index().mul(100).round(2))

print("\nWillingness training distribution:")
print(y_train_willingness.value_counts(normalize=True).sort_index().mul(100).round(2))

# Save model inputs
os.makedirs("../data/model_inputs", exist_ok=True)
X_train.to_csv("../data/model_inputs/X_train.csv", index=False)
X_test.to_csv("../data/model_inputs/X_test.csv", index=False)

y_train_feasibility.to_frame(name=feasibility_target).to_csv(
    "../data/model_inputs/y_train_feasibility.csv",
    index=False
)

y_test_feasibility.to_frame(name=feasibility_target).to_csv(
    "../data/model_inputs/y_test_feasibility.csv",
    index=False
)

y_train_willingness.to_frame(name=willingness_target).to_csv(
    "../data/model_inputs/y_train_willingness.csv",
    index=False
)

y_test_willingness.to_frame(name=willingness_target).to_csv(
    "../data/model_inputs/y_test_willingness.csv",
    index=False
)
```
